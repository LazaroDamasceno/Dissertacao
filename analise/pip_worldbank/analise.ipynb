```python
import polars as pl
import matplotlib.pyplot as plt
```

```python
df_inicial = pl.read_csv('BRA_PovertyRate_20260324_2021_01_02_PROD_2026-09-10.csv')
```

```python
df_inicial.head()
```

```python
df_inicial.columns
```

```python
df_inicial = (
    df_inicial
    .rename({
        'Years': 'ano',
        ' Poverty rate (%)': 'taxa_pobreza_%',
        'Population living in poverty (Million)': 'populacao_na_pobreza_milhoes'
    })
    .drop('Poverty Line')
)
```

```python
df_inicial.head()
```

```python
anos = sorted(df_inicial['ano'].unique().to_list())
print(anos)
```

```python
fig, ax = plt.subplots(
    figsize=(10, 3.5),
    dpi=300,
    layout='constrained'
)

ax.plot(
    df_inicial['ano'],
    df_inicial['taxa_pobreza_%'],
    marker='o'
)

ax.grid(color='grey', linestyle='--', alpha=0.5)
ax.spines[['top', 'right']].set_visible(False)
ax.set_xticks(anos, labels=anos, rotation=90)
ax.set_ylabel('Valor (%)', fontweight='bold')
ax.set_xlabel('Ano', fontweight='bold')

fig.savefig('pip_wb_brasil_taxa_pobreza_%', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(
    figsize=(10, 3.5),
    dpi=300,
    layout='constrained'
)

ax.plot(
    df_inicial['ano'],
    df_inicial['populacao_na_pobreza_milhoes'],
    marker='o'
)

ax.grid(color='grey', linestyle='--', alpha=0.5)
ax.spines[['top', 'right']].set_visible(False)
ax.set_xticks(anos, labels=anos, rotation=90)
ax.set_ylabel('Quantidade (Milhões)', fontweight='bold')
ax.set_xlabel('Ano', fontweight='bold')

fig.savefig('pip_wb_brasil_populacao_na_pobreza_milhoes', dpi=300, bbox_inches='tight')
plt.show()
```
